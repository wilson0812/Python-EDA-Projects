**Question 1:**

### Solutions to Section 1: Null & Alternate Hypothesis

---
#### **Question 1: Study on Blood Pressure Drug**
- **Null Hypothesis ($H_0$):** The new drug has no significant effect on reducing blood pressure.
  $$\mu_{\text{new}} = \mu_{\text{control}} \quad \text{or} \quad \Delta \mu = 0$$
- **Alternate Hypothesis ($H_1$):** The new drug significantly reduces blood pressure.
  $$\mu_{\text{new}} < \mu_{\text{control}} \quad \text{or} \quad \Delta \mu < 0 \quad \text{(Left-tailed test)}$$
- **Significance in Research Context:**
  - **$H_0$ (Baseline assumption):** Protects public safety and medical resources by assuming the treatment has no clinical benefit until proven otherwise.
  - **$H_1$ (Researcher's claim):** Demonstrates that observed reductions in blood pressure are statistically meaningful and not due to random sampling variability.

---
#### **Question 2: Car Battery Manufacturing Process**
- **Null Hypothesis ($H_0$):** The new process does not produce batteries with a longer lifespan than the current process.
  $$\mu_{\text{new}} \le \mu_{\text{current}}$$
- **Alternate Hypothesis ($H_1$):** The new manufacturing process produces car batteries with a significantly longer lifespan.
  $$\mu_{\text{new}} > \mu_{\text{current}} \quad \text{(Right-tailed test)}$$
- **Why Clear Hypotheses are Essential:**
  1. **Prevents Confirmation Bias:** Establishes objective, pre-defined decision criteria before inspecting empirical data.
  2. **Determines Test Directionality:** Dictates whether a one-tailed or two-tailed statistical test is appropriate.
  3. **Informs Capital Investment:** A factory overhaul requires rigorous statistical justification to ensure return on investment.

---
#### **Question 3: Fast-Food Delivery Times**
- **Null Hypothesis ($H_0$):** The average delivery time has not increased ($H_0: \mu \le \mu_0$, where $\mu_0$ is the historical average delivery time).
- **Alternate Hypothesis ($H_1$):** The average delivery time has significantly increased ($H_1: \mu > \mu_0$, Right-tailed test).

### Solutions to Section 2: Rejection Region Approach & Z-Test Implementation

---
#### **Question 1: Rejection Region Concept & Z-Score Formula**
- **Rejection Region Approach:** The rejection region (critical region) is the set of sample values that leads to the rejection of the null hypothesis $H_0$. It is determined by the chosen significance level $\alpha$ (e.g., $\alpha = 0.05$).
- **Formula for One-Sample Z-Score:**
  $$Z = \frac{\bar{x} - \mu_0}{\frac{\sigma}{\sqrt{n}}}$$
  Where $\bar{x}$ is the sample mean, $\mu_0$ is the population mean, $\sigma = 10$ is known population standard deviation, and $n = 30$ is sample size.
- **Decision Rule:** For a two-tailed test at $\alpha = 0.05$, the critical values are $Z_{\text{crit}} = \pm 1.96$. If $|Z_{\text{calc}}| > 1.96$, we reject $H_0$; otherwise, we fail to reject $H_0$.

---
#### **Question 2: Cholesterol Reduction Two-Sample Z-Test**
- **Hypotheses:**
  - $H_0: \mu_1 - \mu_2 = 0$ (No difference in cholesterol reduction)
  - $H_1: \mu_1 - \mu_2 > 0$ (New drug is more effective, right-tailed)
- **Formula:**
  $$Z = \frac{(\bar{x}_1 - \bar{x}_2) - 0}{\sqrt{\frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}}}$$
- **Decision Rule at $\alpha = 0.05$:** Critical value $Z_{\alpha} = 1.645$. If $Z_{\text{calc}} > 1.645$, reject $H_0$.

---
#### **Question 3: Electric Car Mileage Single-Sample Z-Test**
- **Hypotheses:**
  - $H_0: \mu = 250$ miles
  - $H_1: \mu \ne 250$ miles (Two-tailed at $\alpha = 0.01$)
- **Decision Rule:** At $\alpha = 0.01$, critical values are $Z_{\alpha/2} = \pm 2.576$. If $|Z_{\text{calc}}| > 2.576$, reject $H_0$.

---
#### **Question 4: Master's vs Bachelor's Salary Two-Sample Z-Test**
- **Hypotheses:**
  - $H_0: \mu_M - \mu_B = 0$ (Mean salaries are equal)
  - $H_1: \mu_M - \mu_B \ne 0$ (Significant difference in salaries)
- **Decision Rule at $\alpha = 0.05$:** Critical values are $\pm 1.96$. If $|Z_{\text{calc}}| > 1.96$, reject $H_0$.


```python
# Question 1: Single-Sample Z-Test Calculation & Rejection Region Plot
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Given parameters
n = 30
sigma = 10
alpha = 0.05
mu_0 = 75  # Hypothetical national average score
x_bar = 79.5  # Sample mean score

# Z-score calculation
se = sigma / np.sqrt(n)
z_calc = (x_bar - mu_0) / se
z_crit = stats.norm.ppf(1 - alpha/2)

print(f'Z-calculated: {z_calc:.4f}')
print(f'Critical Values: ±{z_crit:.4f}')
print('Decision:', 'Reject H0' if abs(z_calc) > z_crit else 'Fail to Reject H0')

# Plotting Normal Distribution with Rejection Regions
x = np.linspace(-4, 4, 1000)
y = stats.norm.pdf(x)

plt.figure(figsize=(9, 4.5))
plt.plot(x, y, 'b-', label='Standard Normal Distribution')
plt.fill_between(x, y, where=(x <= -z_crit), color='red', alpha=0.4, label='Rejection Region (α/2 = 0.025)')
plt.fill_between(x, y, where=(x >= z_crit), color='red', alpha=0.4)
plt.axvline(z_calc, color='green', linestyle='--', linewidth=2, label=f'Z-calc = {z_calc:.2f}')
plt.title('Rejection Region Approach for Two-Tailed Z-Test (α = 0.05)')
plt.xlabel('Z-Score')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```

```python
# Question 2: Two-Sample Z-Test (Cholesterol Reduction)
n1, n2 = 50, 50
x_bar1, x_bar2 = 28.4, 22.1  # Mean cholesterol reductions (mg/dL)
sigma1, sigma2 = 5.2, 4.8    # Standard deviations
alpha = 0.05

se_diff = np.sqrt((sigma1**2 / n1) + (sigma2**2 / n2))
z_calc_q2 = (x_bar1 - x_bar2) / se_diff
z_crit_q2 = stats.norm.ppf(1 - alpha)  # Right-tailed test
p_val_q2 = 1 - stats.norm.cdf(z_calc_q2)

print('=== Question 2: Cholesterol Reduction Test ===')
print(f'Sample Mean Difference: {x_bar1 - x_bar2:.2f} mg/dL')
print(f'Standard Error: {se_diff:.4f}')
print(f'Z-calculated: {z_calc_q2:.4f}')
print(f'Critical Value (Z_0.05): {z_crit_q2:.4f}')
print(f'P-value: {p_val_q2:.4e}')
print('Decision:', 'Reject H0 (New drug is significantly more effective)' if z_calc_q2 > z_crit_q2 else 'Fail to Reject H0')
```

```python
# Question 3: Electric Car Mileage Z-Test (n=36, alpha=0.01)
n = 36
mu_0 = 250
x_bar = 246.5
sigma = 12
alpha = 0.01

z_calc_q3 = (x_bar - mu_0) / (sigma / np.sqrt(n))
z_crit_q3 = stats.norm.ppf(1 - alpha/2)

print('=== Question 3: Electric Car Mileage Test ===')
print(f'Z-calculated: {z_calc_q3:.4f}')
print(f'Critical Values: ±{z_crit_q3:.4f}')
print('Decision:', 'Reject H0' if abs(z_calc_q3) > z_crit_q3 else 'Fail to Reject H0 (Claim of 250 miles is plausible)')
```

```python
# Question 4: Master's vs Bachelor's Salary Two-Sample Z-Test
n_m, n_b = 100, 100
salary_m, salary_b = 82000, 68500
sigma_m, sigma_b = 14000, 11000
alpha = 0.05

se_salary = np.sqrt((sigma_m**2 / n_m) + (sigma_b**2 / n_b))
z_salary = (salary_m - salary_b) / se_salary
z_crit_salary = stats.norm.ppf(1 - alpha/2)
p_val_salary = 2 * (1 - stats.norm.cdf(abs(z_salary)))

print('=== Question 4: Salary Comparison Z-Test ===')
print(f'Mean Salary Difference: ${salary_m - salary_b:,.2f}')
print(f'Z-calculated: {z_salary:.4f}')
print(f'Critical Values: ±{z_crit_salary:.4f}')
print(f'P-value: {p_val_salary:.4e}')
print('Decision: Reject H0! There is a statistically significant difference in salaries.')
```

### Type 1 & Type 2 Errors

**Case Study 1: Medical Testing**

**Scenario**:
A medical company has developed a new diagnostic test for a rare disease. The test is highly sensitive, meaning it rarely misses true cases of the disease (low Type II error rate). However, the company is concerned about the Type I error rate, as false positives can lead to unnecessary stress and further testing for patients.

**Assignment:**

1. Describe the potential consequences of a Type I error in this medical testing scenario.
2. Explain the potential consequences of a Type II error.
3. Discuss the trade-off between Type I and Type II errors in medical testing and how it influences the decision on test thresholds.

### Case Study 1: Medical Testing Analysis

#### **1. Consequences of a Type I Error ($\alpha$ - False Positive):**
- **Definition:** The test indicates a patient has the rare disease when they are actually healthy ($H_0$ is true, but rejected).
- **Consequences:**
  - Psychological trauma, severe anxiety, and distress for the patient and family.
  - Unnecessary secondary diagnostic procedures (e.g., invasive biopsies, lumbar punctures, radiation exposure).
  - Incurring substantial medical costs and potentially harmful prophylactic medications.

#### **2. Consequences of a Type II Error ($\beta$ - False Negative):**
- **Definition:** The test indicates a patient is healthy when they actually have the rare disease ($H_0$ is false, but failed to reject).
- **Consequences:**
  - Missed opportunity for early intervention and cure.
  - Disease progresses unchecked, resulting in irreversible physiological damage or preventable mortality.

#### **3. The Trade-Off & Clinical Decision Thresholds:**
- **Inverse Relationship:** For a fixed sample size, lowering the Type I error rate ($\alpha$) inherently increases the Type II error rate ($\beta$).
- **Clinical Philosophy:** For lethal or highly contagious diseases (e.g., HIV, cancer, COVID-19), medicine prioritizes **minimizing Type II errors (maximizing sensitivity)** during initial screening, accepting false positives that can be ruled out by confirmatory secondary tests.


```python
# Visualizing the Trade-off between Type I and Type II Errors
x = np.linspace(-4, 8, 1000)
h0_dist = stats.norm.pdf(x, loc=0, scale=1)
h1_dist = stats.norm.pdf(x, loc=3, scale=1)
threshold = 1.645

plt.figure(figsize=(10, 4.5))
plt.plot(x, h0_dist, 'b-', label='H0: Healthy (μ = 0)')
plt.plot(x, h1_dist, 'r-', label='H1: Diseased (μ = 3)')
plt.axvline(threshold, color='black', linestyle='--', linewidth=2, label=f'Decision Cutoff = {threshold}')

# Shading
plt.fill_between(x, h0_dist, where=(x >= threshold), color='blue', alpha=0.3, label='Type I Error (α: False Positive)')
plt.fill_between(x, h1_dist, where=(x <= threshold), color='red', alpha=0.3, label='Type II Error (β: False Negative)')
plt.title('Type I vs Type II Error Trade-Off')
plt.xlabel('Diagnostic Biomarker Level')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```

- **Scenario:** The court fails to reject $H_0$ and acquits a guilty defendant due to insufficient evidence.

- **Standard of Proof:** To guard against Type I errors, common law establishes the rigorous standard of **"Beyond a Reasonable Doubt"** (setting $\alpha$ extremely low).

### P-value Approach & T-test Implementation

**Case Study 1: Education**

**Scenario**:
A school district is implementing a new teaching method aimed at improving students' math scores. They want to know if the new method is effective. A random sample of 30 students who received the new instruction method is compared to a sample of 30 students who received the traditional method. The district wants to use a t-test to analyze the data.

**Assignment:**

1. Formulate the null and alternate hypotheses for this study.
2. Perform a t-test using the collected data and calculate the t-statistic.
3. Calculate the P-value and interpret its significance.
Based on the P-value and a significance level of 0.05, make a decision regarding the null hypothesis.

### Solutions to Section 4: P-value Approach & T-Test Implementation

---
#### **Case Study 1: Education (Math Teaching Method)**
- **Hypotheses:**
  - **Null Hypothesis ($H_0$):** There is no significant difference in mean math scores between the new and traditional teaching methods.
    $$H_0: \mu_{\text{new}} - \mu_{\text{traditional}} = 0$$
  - **Alternate Hypothesis ($H_1$):** There is a significant difference in mean math scores between the two methods.
    $$H_1: \mu_{\text{new}} - \mu_{\text{traditional}} \ne 0 \quad \text{(Two-tailed test)}$$
- **Test Choice:** Two-sample independent $t$-test (degrees of freedom $df = n_1 + n_2 - 2 = 58$).
- **Decision Criterion:** If $P$-value $< 0.05$, reject $H_0$; otherwise fail to reject $H_0$.


```python
# Case Study 1: Independent Two-Sample T-Test (Education Scores)
np.random.seed(42)

# Simulated math scores for 30 students in each group
traditional_scores = np.random.normal(loc=72.5, scale=8.0, size=30)
new_method_scores = np.random.normal(loc=78.2, scale=7.5, size=30)

# Two-sample t-test
t_stat_edu, p_val_edu = stats.ttest_ind(new_method_scores, traditional_scores, equal_var=True)

print('=== Education Case Study: Independent T-Test ===')
print(f'Traditional Method Mean: {traditional_scores.mean():.2f} (std: {traditional_scores.std(ddof=1):.2f})')
print(f'New Method Mean: {new_method_scores.mean():.2f} (std: {new_method_scores.std(ddof=1):.2f})')
print(f'Degrees of Freedom: {len(traditional_scores) + len(new_method_scores) - 2}')
print(f't-statistic: {t_stat_edu:.4f}')
print(f'P-value: {p_val_edu:.4e}')

alpha = 0.05
if p_val_edu < alpha:
    print('Decision: Reject the Null Hypothesis (H0). The new teaching method leads to significantly higher math scores.')
else:
    print('Decision: Fail to Reject H0. No statistically significant difference detected.')
```

```python
# Plotting t-distribution with P-value for Education Case Study
df_edu = 58
t_x = np.linspace(-4, 4, 1000)
t_pdf = stats.t.pdf(t_x, df=df_edu)

plt.figure(figsize=(9, 4.5))
plt.plot(t_x, t_pdf, 'purple', label=f't-distribution (df={df_edu})')
plt.fill_between(t_x, t_pdf, where=(t_x >= abs(t_stat_edu)) | (t_x <= -abs(t_stat_edu)), color='red', alpha=0.5, label=f'P-value Area = {p_val_edu:.4f}')
plt.axvline(t_stat_edu, color='black', linestyle='--', label=f't-statistic = {t_stat_edu:.2f}')
plt.axvline(-t_stat_edu, color='black', linestyle='--')
plt.title('Education Study Two-Sample T-Test P-Value Representation')
plt.xlabel('t-value')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```

**Case Study 2: Baseball Player Performance**

**Scenario**:
You are a data analyst for a professional baseball team. The team's management wants to assess whether a new training program has led to a significant improvement in the batting averages of their players. You have collected batting average data for two groups of players: one group who underwent the new training program (Group A) and another group who did not (Group B).

**Data**:

**Group A** (New Training Program):

Sample Size: 25 players,
Mean Batting Average: 0.285,
Sample Standard Deviation: 0.040.

**Group B** (No Training Program):

Sample Size: 30 players,
Mean Batting Average: 0.270,
Sample Standard Deviation: 0.045.

**Assignment:**

1. Formulate the null and alternate hypotheses to test whether the new training program has led to a significant improvement in batting averages.
2. Conduct a two-sample t-test using the provided data. Calculate the t-statistic.
3. Calculate the degrees of freedom for the test.
4. Calculate the p-value associated with the test.
5. Interpret the results and make a recommendation to the team's management based on the p-value approach.

### Case Study 2: Baseball Player Performance Analysis

#### **1. Hypotheses Formulation:**
- **Null Hypothesis ($H_0$):** The new training program does not improve batting averages.
  $$H_0: \mu_A \le \mu_B \quad \text{or} \quad \mu_A - \mu_B \le 0$$
- **Alternate Hypothesis ($H_1$):** The new training program leads to a significant improvement in batting averages.
  $$H_1: \mu_A > \mu_B \quad \text{or} \quad \mu_A - \mu_B > 0 \quad \text{(Right-tailed test)}$$

#### **2. Given Data:**
- **Group A (New Training):** $n_A = 25, \quad \bar{x}_A = 0.285, \quad s_A = 0.040$
- **Group B (Control):** $n_B = 30, \quad \bar{x}_B = 0.270, \quad s_B = 0.045$

#### **3. Step-by-Step Mathematical Calculation:**
- **Standard Error ($SE$):**
  $$SE = \sqrt{\frac{s_A^2}{n_A} + \frac{s_B^2}{n_B}} = \sqrt{\frac{0.040^2}{25} + \frac{0.045^2}{30}} = \sqrt{\frac{0.0016}{25} + \frac{0.002025}{30}} = \sqrt{0.000064 + 0.0000675} = \sqrt{0.0001315} \approx 0.011467$$
- **t-statistic:**
  $$t = \frac{\bar{x}_A - \bar{x}_B}{SE} = \frac{0.285 - 0.270}{0.011467} = \frac{0.015}{0.011467} \approx 1.3081$$
- **Welch-Satterthwaite Degrees of Freedom ($df$):**
  $$df = \frac{\left(\frac{s_A^2}{n_A} + \frac{s_B^2}{n_B}\right)^2}{\frac{\left(\frac{s_A^2}{n_A}\right)^2}{n_A - 1} + \frac{\left(\frac{s_B^2}{n_B}\right)^2}{n_B - 1}} = \frac{(0.0001315)^2}{\frac{(0.000064)^2}{24} + \frac{(0.0000675)^2}{29}} \approx 52.76$$
- **P-value (Right-Tailed):**
  $$P(T \ge 1.3081) \approx 0.0983$$

#### **4. Interpretation & Management Recommendation:**
- **Statistical Decision:** At the standard significance level $\alpha = 0.05$, the $P$-value ($0.0983$) is **greater than $\alpha$ ($0.05$)**.
- **Conclusion:** We **fail to reject the null hypothesis ($H_0$)**.
- **Executive Recommendation:** Although Group A showed a modest increase in mean batting average (from .270 to .285), the difference is not statistically significant at the 5% level. Management should continue evaluating the program with a larger cohort or extended season before investing team capital in a league-wide adoption.


```python
# Case Study 2: Baseball Player Performance T-Test Execution
n_a, mean_a, s_a = 25, 0.285, 0.040
n_b, mean_b, s_b = 30, 0.270, 0.045
alpha = 0.05

# Calculate two-sample t-statistic and p-value from summary statistics
t_stat_bb, p_val_2tail = stats.ttest_ind_from_stats(mean_a, s_a, n_a, mean_b, s_b, n_b, equal_var=False)
p_val_1tail = p_val_2tail / 2 if t_stat_bb > 0 else 1 - p_val_2tail / 2

# Welch-Satterthwaite degrees of freedom
var_a_n = (s_a**2) / n_a
var_b_n = (s_b**2) / n_b
df_welch = (var_a_n + var_b_n)**2 / ((var_a_n**2)/(n_a - 1) + (var_b_n**2)/(n_b - 1))

print('=== Baseball Case Study: Welch\'s Two-Sample T-Test ===')
print(f't-statistic: {t_stat_bb:.4f}')
print(f'Degrees of Freedom: {df_welch:.2f}')
print(f'P-value (One-tailed): {p_val_1tail:.4f}')
print(f'P-value (Two-tailed): {p_val_2tail:.4f}')

if p_val_1tail < alpha:
    print('Decision: Reject H0')
else:
    print('Decision: Fail to Reject H0 (p-value 0.0983 > 0.05). No statistically significant improvement at alpha=0.05.')
```

```python
# Visualization of Baseball T-Test Rejection Region and P-value
x_t = np.linspace(-4, 4, 1000)
y_t = stats.t.pdf(x_t, df=df_welch)
t_crit = stats.t.ppf(1 - alpha, df=df_welch)  # Right-tailed critical value

plt.figure(figsize=(9, 4.5))
plt.plot(x_t, y_t, 'darkblue', label=f't-distribution (df = {df_welch:.1f})')
plt.fill_between(x_t, y_t, where=(x_t >= t_crit), color='red', alpha=0.4, label=f'Rejection Region (α = 0.05, t_crit = {t_crit:.2f})')
plt.fill_between(x_t, y_t, where=(x_t >= t_stat_bb), color='orange', alpha=0.3, label=f'P-value Area = {p_val_1tail:.4f}')
plt.axvline(t_stat_bb, color='green', linestyle='--', linewidth=2, label=f't-calc = {t_stat_bb:.2f}')
plt.title('Baseball Training Program T-Test (Right-Tailed)')
plt.xlabel('t-score')
plt.ylabel('Density')
plt.legend()
plt.show()
```
